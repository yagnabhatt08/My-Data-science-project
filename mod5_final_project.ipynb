Use `Pandas` to load the data available in the links above to dataframes. Use these dataframes to load data on to the database `FinalDB.db` as required tables.


```python
import pandas as pd
import sqlite3
import numpy as np

# Define the URLs for the Chicago data files
urls = {
    'ChicagoCensusData': 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCensusData.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01',
    'ChicagoPublicSchools': 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoPublicSchools.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01',
    'ChicagoCrimeData': 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCrimeData.csv?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDB0201ENSkillsNetwork20127838-2021-01-01'
}

# Create a dictionary to store DataFrames
dataframes = {}

print("Loading data from URLs...")
print("=" * 60)

# Load each dataset into pandas DataFrames
for name, url in urls.items():
    print(f"Loading {name}...")
    try:
        # Load CSV from URL
        df = pd.read_csv(url)
        dataframes[name] = df
        
        # Display basic information
        print(f"  ✓ Successfully loaded {name}")
        print(f"    Shape: {df.shape} (rows: {df.shape[0]}, columns: {df.shape[1]})")
        print(f"    Columns: {list(df.columns[:5])}...")  # Show first 5 columns
        print(f"    First few rows:")
        print(df.head(3).to_string(index=False))
        print("-" * 40)
        
    except Exception as e:
        print(f"  ✗ Error loading {name}: {e}")

# Display summary of loaded data
print("\n" + "=" * 60)
print("SUMMARY OF LOADED DATA")
print("=" * 60)

for name, df in dataframes.items():
    print(f"\n{name}:")
    print(f"  Total rows: {len(df):,}")
    print(f"  Total columns: {len(df.columns)}")
    print(f"  Column names:")
    for i, col in enumerate(df.columns, 1):
        print(f"    {i:2}. {col}")
    
    # Show data types
    print(f"  Data types:")
    dtypes_summary = df.dtypes.value_counts()
    for dtype, count in dtypes_summary.items():
        print(f"    {dtype}: {count} columns")
    
    # Check for missing values
    missing_values = df.isnull().sum().sum()
    if missing_values > 0:
        print(f"  Missing values: {missing_values:,} total")
        cols_with_missing = df.isnull().sum()
        cols_with_missing = cols_with_missing[cols_with_missing > 0]
        print(f"  Columns with missing values: {len(cols_with_missing)}")

# Step 2: Load data into SQLite database
print("\n" + "=" * 60)
print("LOADING DATA INTO SQLite DATABASE: FinalDB.db")
print("=" * 60)

# Connect to SQLite database
conn = sqlite3.connect('FinalDB.db')
cursor = conn.cursor()

# Function to clean column names for SQL compatibility
def clean_column_names(df):
    """Clean column names for SQL compatibility"""
    df_clean = df.copy()
    # Replace spaces and special characters with underscores
    df_clean.columns = df_clean.columns.str.replace(' ', '_').str.replace('-', '_').str.replace('(', '').str.replace(')', '').str.replace('/', '_').str.replace('%', 'pct').str.replace('$', '').str.replace('.', '').str.lower()
    return df_clean

# Clean column names for each DataFrame
for name in dataframes.keys():
    dataframes[name] = clean_column_names(dataframes[name])
    print(f"Cleaned column names for {name}")

# Load each DataFrame into the database
for name, df in dataframes.items():
    table_name = name.lower()
    
    try:
        # Create or replace table in database
        df.to_sql(table_name, conn, if_exists='replace', index=False)
        
        # Verify data was loaded
        cursor.execute(f"SELECT COUNT(*) FROM {table_name}")
        count = cursor.fetchone()[0]
        
        # Get column information
        cursor.execute(f"PRAGMA table_info({table_name})")
        columns = cursor.fetchall()
        column_names = [col[1] for col in columns]
        
        print(f"\n✓ Successfully loaded '{table_name}' table:")
        print(f"  Rows loaded: {count:,}")
        print(f"  Columns: {len(columns)}")
        print(f"  Column names (first 5): {column_names[:5]}")
        
        # Show sample data
        if count > 0:
            sample_query = f"SELECT * FROM {table_name} LIMIT 2"
            sample_df = pd.read_sql_query(sample_query, conn)
            print(f"  Sample data:")
            print(sample_df.to_string(index=False))
            
    except Exception as e:
        print(f"\n✗ Error loading '{table_name}': {e}")

# Display all tables in the database
print("\n" + "=" * 60)
print("DATABASE SCHEMA SUMMARY")
print("=" * 60)

# List all tables in the database
cursor.execute("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;")
tables = cursor.fetchall()

print(f"\nTables in 'FinalDB.db' database:")
for table in tables:
    table_name = table[0]
    
    # Get row count for each table
    cursor.execute(f"SELECT COUNT(*) FROM {table_name}")
    row_count = cursor.fetchone()[0]
    
    # Get column information
    cursor.execute(f"PRAGMA table_info({table_name})")
    columns = cursor.fetchall()
    
    print(f"\n{table_name.upper()}:")
    print(f"  Rows: {row_count:,}")
    print(f"  Columns: {len(columns)}")
    print("  Column names:")
    for col in columns:
        col_name, col_type = col[1], col[2]
        print(f"    - {col_name} ({col_type})")

# Optional: Create some example queries to demonstrate the data
print("\n" + "=" * 60)
print("EXAMPLE QUERIES")
print("=" * 60)

# Example 1: Census data summary
try:
    census_query = """
    SELECT 
        COUNT(*) as total_communities,
        AVG(per_capita_income) as avg_per_capita_income,
        SUM(population) as total_population
    FROM chicagocensusdata
    """
    census_summary = pd.read_sql_query(census_query, conn)
    print("\n1. Chicago Census Data Summary:")
    print(census_summary.to_string(index=False))
except:
    print("\n1. Could not run census query")

# Example 2: School data summary
try:
    schools_query = """
    SELECT 
        COUNT(*) as total_schools,
        COUNT(DISTINCT community_area_name) as communities_with_schools,
        AVG(average_student_attendance) as avg_attendance
    FROM chicagopublicschools
    """
    schools_summary = pd.read_sql_query(schools_query, conn)
    print("\n2. Chicago Public Schools Summary:")
    print(schools_summary.to_string(index=False))
except:
    print("\n2. Could not run schools query")

# Example 3: Crime data summary
try:
    crime_query = """
    SELECT 
        COUNT(*) as total_crimes,
        COUNT(DISTINCT primary_type) as crime_types,
        MIN(date) as earliest_date,
        MAX(date) as latest_date
    FROM chicagocrimedata
    """
    crime_summary = pd.read_sql_query(crime_query, conn)
    print("\n3. Chicago Crime Data Summary:")
    print(crime_summary.to_string(index=False))
except:
    print("\n3. Could not run crime query")

# Commit changes and close connection
conn.commit()
conn.close()

print("\n" + "=" * 60)
print("DATABASE CREATION COMPLETE")
print("=" * 60)
print(f"✓ All data successfully loaded into 'FinalDB.db'")
print(f"✓ Total tables created: {len(tables)}")
print(f"✓ Total rows across all tables: {sum([len(dataframes[name]) for name in dataframes])}")
print("\nTables available:")
for table in tables:
    print(f"  - {table[0]}")
print("\nYou can now query the database using:")
print("  conn = sqlite3.connect('FinalDB.db')")
print("  df = pd.read_sql_query('SELECT * FROM table_name', conn)")
```

```python
import pandas as pd
import sqlite3

# URL for Chicago Census Data
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCensusData.csv"

# Load data from URL
df = pd.read_csv(url)

# Clean column names for SQL
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '').str.replace('-', '_')

# Connect to database and save
conn = sqlite3.connect('FinalDB.db')
df.to_sql('chicagocensusdata', conn, if_exists='replace', index=False)

print(f"✅ Loaded {len(df)} rows into 'chicagocensusdata' table")
print(f"✅ Columns: {list(df.columns)}")

# Test query
query = "SELECT community_area_number, community_area_name, per_capita_income FROM chicagocensusdata WHERE per_capita_income < 11000"
result = pd.read_sql_query(query, conn)
print(f"\n✅ Found {len(result)} areas with per capita income < 11000")
print(result.head())

conn.close()
```

### Problem 4

##### List all kidnapping crimes involving a child?


```python
# One-liner to get all case numbers for kidnapping crimes involving a child
%sql SELECT case_number FROM chicagocrimedata WHERE (UPPER(primary_type) LIKE '%KIDNAP%' OR UPPER(description) LIKE '%KIDNAP%') AND UPPER(description) LIKE '%CHILD%' ORDER BY case_number;
```

```python
# Save all case numbers to a pandas DataFrame
import pandas as pd
import sqlite3

# Connect to database
conn = sqlite3.connect('FinalDB.db')

# Query to get kidnapping crimes involving children
query = """
SELECT DISTINCT case_number
FROM chicagocrimedata
WHERE (
    UPPER(primary_type) LIKE '%KIDNAP%' OR
    UPPER(primary_type) LIKE '%ABDUCT%' OR
    UPPER(description) LIKE '%KIDNAP%' OR
    UPPER(description) LIKE '%ABDUCT%'
)
AND (
    UPPER(description) LIKE '%CHILD%' OR
    UPPER(description) LIKE '%MINOR%' OR
    UPPER(description) LIKE '%JUVENILE%' OR
    UPPER(description) LIKE '%INFANT%' OR
    UPPER(description) LIKE '%BABY%' OR
    UPPER(description) LIKE '%TODDLER%' OR
    UPPER(description) LIKE '%BOY%' OR
    UPPER(description) LIKE '%GIRL%'
)
ORDER BY case_number
"""

# Execute query
kidnapping_child_cases = pd.read_sql_query(query, conn)
conn.close()

# Display results
print(f"Found {len(kidnapping_child_cases)} kidnapping crimes involving a child")
print("\nCase Numbers:")
if len(kidnapping_child_cases) > 0:
    print(kidnapping_child_cases['case_number'].to_string(index=False))
else:
    print("No kidnapping crimes involving children found.")
```
